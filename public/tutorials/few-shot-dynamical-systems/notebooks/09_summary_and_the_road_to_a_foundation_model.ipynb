# 09 · Summary and the Road to a Foundation Model

Eight chapters ago we had a family of nonlinear oscillators and a learner that could only choose an
input and read a noisy position. Along the way, a two-number context, learned from 20 other
systems, turned out to identify new systems from a few seconds of data, tell us which experiment to
run, warn us when it no longer applied, and control a system it had never seen. This closing chapter
steps back and maps every piece onto its counterpart in a real, large-scale foundation model for
dynamical systems.

| Step | In this course | At scale |
|---|---|---|
| Model class | known physics equation, 4 coefficients | structured neural state-space model / sequence model |
| Pretraining | per-system identification + noise-whitened PCA (chapter 03) | joint training over many heterogeneous systems |
| Task context | 2 numbers $z$ with a Gaussian prior (chapter 04) | learned context encoder, low-rank adapters, hypernetworks |
| Adaptation | MAP with Laplace uncertainty (chapter 04) | amortised inference or fast gradient adaptation |
| Experiment design | D-optimal selection among candidates (chapter 06) | safe, constrained, sequential (Bayesian) design |
| Reliability | misfit test against a calibrated threshold (chapter 07) | calibrated uncertainty and out-of-distribution detection |
| Use | feedforward tracking (chapter 08) | MPC, reinforcement learning, safe learning control |

---
## 1. What we learned

1. Knowledge of *similar* systems turns a four-parameter identification problem into a
   two-parameter one. This matters most exactly when data is scarce (chapter 05).
2. The value of data depends on the experiment. An information criterion computed from the
   pretrained model chooses experiments that swing the system through its informative regime
   (chapter 06).
3. Shared knowledge is a double-edged sword: outside the family it biases the answer. The same
   probabilistic machinery can detect this (chapter 07).
4. Better models translate into better control (chapter 08).

---
## 2. Open questions this toy problem raises

- *Identifiability:* when can the context of a learned (neural) backbone be recovered from partial,
  noisy measurements, and how much data does it need?
- *Calibration:* the Laplace posterior assumes white noise, while real disturbances are coloured.
  How do we obtain uncertainty we can trust for decision-making?
- *Safe excitation:* informative experiments push systems hard. How do we trade information
  against state and input constraints, ideally with guarantees?
- *Adapter capacity:* how do we decide online when a small context suffices and when the model
  must grow?
- *Closing the loop:* how should identification, uncertainty and control be designed jointly?
